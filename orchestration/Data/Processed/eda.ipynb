```python
import pandas as pd

# Load dataset (replace with your filename)
df = pd.read_csv("tickets_inputs_eng2.csv")

# Inspect first rows
df.head()
```

```python

# Suppose your target column is called "label"
label_col = "relevant_topics"

# Count class distribution
class_counts = df[label_col].value_counts()
print("\nClass distribution:")
print(class_counts)

# Check balance ratio
ratio = class_counts.max() / class_counts.min()
print(f"\nBalance ratio (max/min): {ratio:.2f}")

if ratio <= 1.5:
    print("✅ Dataset is relatively balanced")
else:
    print("⚠️ Dataset is imbalanced")
```
